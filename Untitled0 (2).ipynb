## Analyze and Visualize Class Distribution of MIT-BIH Dataset
The last column (index 187) contains the labels. Let's load both train and test datasets, plot sample signals for each class, and analyze class imbalance.

```python
import numpy as np
import matplotlib.pyplot as plt

train_df = pd.read_csv('/content/ecg_data/mitbih_train.csv', header=None)
test_df = pd.read_csv('/content/ecg_data/mitbih_test.csv', header=None)

# Class mapping for MIT-BIH
class_map = {
    0.0: "Normal",
    1.0: "Supraventricular ectopic beat",
    2.0: "Ventricular ectopic beat",
    3.0: "Fusion beat",
    4.0: "Unknown beat"
}

# Check class distributions
print("Train class distribution:")
print(train_df[187].value_counts().rename(index=class_map))

print("\nTest class distribution:")
print(test_df[187].value_counts().rename(index=class_map))

# Plot one sample of each class
plt.figure(figsize=(15, 8))
for label, name in class_map.items():
    sample = train_df[train_df[187] == label].iloc[0, :187]
    plt.plot(sample, label=f"Class {int(label)}: {name}")

plt.title("Sample ECG signals for each class in MIT-BIH dataset")
plt.xlabel("Time steps")
plt.ylabel("Normalized amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

## Build and Train a 1D CNN Classifier with Class Weights
Since ECG signals are sequential time series, 1D Convolutional Neural Networks (CNN) are highly effective at capturing local temporal features. Let's preprocess the inputs, compute balanced class weights, define our architecture, and train the model.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import numpy as np
import pandas as pd
import zipfile
import os

# 1. Safely handle ZIP extraction with correct path
zip_path = '/content/archive.zip'
extract_path = '/content/ecg_data'

if not os.path.exists(extract_path) or len(os.listdir(extract_path)) == 0:
    print(f"Extracting dataset from {zip_path} to {extract_path}...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Extraction complete.")

# Load datasets
train_df = pd.read_csv('/content/ecg_data/mitbih_train.csv', header=None)
test_df = pd.read_csv('/content/ecg_data/mitbih_test.csv', header=None)

# 2. Prepare Features and Labels
X_train = train_df.iloc[:, :187].values
y_train = train_df.iloc[:, 187].values.astype(int)

X_test = test_df.iloc[:, :187].values
y_test = test_df.iloc[:, 187].values.astype(int)

# Reshape for 1D CNN: (samples, timesteps, channels)
X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)

# Compute class weights to handle imbalance
classes = np.unique(y_train)
class_weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
# Cast keys and values explicitly to standard python types to avoid tensor conversion errors
class_weight_dict = {int(c): float(w) for c, w in zip(classes, class_weights)}
print("Calculated Class Weights:", class_weight_dict)

# 3. Build 1D CNN Model
def build_model(input_shape, num_classes):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        layers.Conv1D(64, kernel_size=6, activation='relu'),
        layers.BatchNormalization(),
        layers.MaxPool1D(pool_size=3, strides=2),

        layers.Conv1D(128, kernel_size=5, activation='relu'),
        layers.BatchNormalization(),
        layers.MaxPool1D(pool_size=3, strides=2),

        layers.Conv1D(256, kernel_size=3, activation='relu'),
        layers.BatchNormalization(),
        layers.MaxPool1D(pool_size=3, strides=2),

        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model

model = build_model(input_shape=(187, 1), num_classes=5)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()

# 4. Train the Model
history = model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.15,
    class_weight=class_weight_dict,
    verbose=1
)
```

## Evaluate the Model
Let's evaluate the performance on the test set, display the classification report, and plot a confusion matrix to observe how well the model distinguishes each category.

```python
# Predict classes on test set
y_pred_prob = model.predict(X_test)
y_pred = np.argmax(y_pred_prob, axis=1)

# Classification Report
print("\nClassification Report:")
target_names = [class_map[i] for i in sorted(class_map.keys())]
print(classification_report(y_test, y_pred, target_names=target_names))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=target_names, yticklabels=target_names, cmap='Blues')
plt.title("ECG Arrhythmia Classification - Confusion Matrix")
plt.ylabel("True Class")
plt.xlabel("Predicted Class")
plt.tight_layout()
plt.show()
```
